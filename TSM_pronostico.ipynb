```python
# -*- coding: utf-8 -*-

# ============================================================
# PRONÓSTICO SEMANAL DE TEMPERATURA SUPERFICIAL DEL MAR
# ECMWF IFS ENS - 46 DÍAS
#
# Área:
#   90°W - 68°W
#   22°S - 2°N
#
# Metodología:
#   - Última inicialización disponible
#   - Se descartan las primeras 48 horas
#   - 6 semanas completas de 7 días
#   - Promedio temporal semanal por miembro
#   - Promedio de los 101 miembros
#
# Isotermas:
#   26.0 °C
#   27.0 °C
#   28.5 °C
#   29.0 °C
# ============================================================


# ============================================================
# 0. INSTALAR PAQUETES - GOOGLE COLAB
# ============================================================

!uv pip install -q dynamical-catalog cartopy netcdf4


# ============================================================
# 1. LIBRERÍAS
# ============================================================

import os

import numpy as np
import pandas as pd
import xarray as xr

import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

import cartopy.crs as ccrs
import cartopy.feature as cfeature

from matplotlib.lines import Line2D

import dynamical_catalog


# ============================================================
# 2. CONFIGURACIÓN GENERAL
# ============================================================

CATALOGO = (
    "ecmwf-ifs-ens-forecast-46-day-daily-1-5-degree"
)

VAR_SST = "sea_surface_temperature"


# ============================================================
# CONFIGURACIÓN TEMPORAL
# ============================================================

RETRASO_DIAS = 2

N_SEMANAS = 6

DIAS_SEMANA = 7


# ============================================================
# DOMINIO
# ============================================================

LON_MIN = -90
LON_MAX = -68

LAT_MIN = -22
LAT_MAX = 2


# ============================================================
# DIRECTORIO DE SALIDA
# ============================================================

OUTPUT_DIR = "/content/salidas_tsm"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)


# ============================================================
# 3. ABRIR ECMWF IFS ENS
# ============================================================

print("=" * 100)
print("ABRIENDO ECMWF IFS ENS")
print("=" * 100)

ds = dynamical_catalog.open(
    CATALOGO,
    chunks=None
)

print("\nDataset abierto correctamente.")

print("\nDimensiones:")
print(ds.sizes)


# ============================================================
# 4. VERIFICAR VARIABLE SST
# ============================================================

if VAR_SST not in ds.data_vars:

    raise KeyError(
        f"No existe la variable {VAR_SST}"
    )


print("\nVariable SST:")
print(ds[VAR_SST])


# ============================================================
# 5. ÚLTIMA INICIALIZACIÓN DISPONIBLE
# ============================================================

INIT_TIME = ds.init_time.values[-1]

INIT = pd.Timestamp(
    INIT_TIME
)


print("\n" + "=" * 100)
print("INICIALIZACIÓN")
print("=" * 100)

print(
    "Inicialización ECMWF :",
    INIT
)

print(
    "Miembros disponibles :",
    ds.sizes["ensemble_member"]
)

print(
    "Unidad SST           :",
    ds[VAR_SST].attrs.get(
        "units",
        "sin unidad"
    )
)


# ============================================================
# 6. DEFINIR VENTANA OPERATIVA
#
# Se eliminan las primeras 48 horas.
#
# Semana 1 = D+2 ... D+8
# Semana 2 = D+9 ... D+15
# ...
# Semana 6 = D+37 ... D+43
# ============================================================

LEAD_INICIO = pd.Timedelta(
    days=RETRASO_DIAS
)

LEAD_FIN = (
    LEAD_INICIO
    +
    pd.Timedelta(
        days=(
            N_SEMANAS
            *
            DIAS_SEMANA
            - 1
        )
    )
)


FECHA_INICIO = (
    INIT
    +
    LEAD_INICIO
)

FECHA_FIN = (
    INIT
    +
    LEAD_FIN
)


print("\n" + "=" * 100)
print("VENTANA OPERATIVA")
print("=" * 100)

print(
    "Inicialización :",
    INIT
)

print(
    "Inicio (+48 h) :",
    FECHA_INICIO
)

print(
    "Final          :",
    FECHA_FIN
)

print(
    "Lead inicial   :",
    LEAD_INICIO
)

print(
    "Lead final     :",
    LEAD_FIN
)


# ============================================================
# 7. EXTRAER SST
# ============================================================

print("\n" + "=" * 100)
print("EXTRAYENDO TEMPERATURA SUPERFICIAL DEL MAR")
print("=" * 100)


sst = (
    ds[VAR_SST]
    .sel(
        init_time=INIT_TIME
    )
    .sel(
        lead_time=slice(
            LEAD_INICIO,
            LEAD_FIN
        )
    )
)


# ============================================================
# 8. RECORTAR LONGITUD
# ============================================================

sst = sst.sel(
    longitude=slice(
        LON_MIN,
        LON_MAX
    )
)


# ============================================================
# 9. RECORTAR LATITUD
#
# Se controla automáticamente si latitude viene
# ascendente o descendente.
# ============================================================

if (
    float(sst.latitude.values[0])
    >
    float(sst.latitude.values[-1])
):

    sst = sst.sel(
        latitude=slice(
            LAT_MAX,
            LAT_MIN
        )
    )

else:

    sst = sst.sel(
        latitude=slice(
            LAT_MIN,
            LAT_MAX
        )
    )


# ============================================================
# 10. CARGAR DATOS
# ============================================================

print("\nCargando datos...")

sst = sst.load()

print("Datos cargados correctamente.")


print("\nDimensiones finales:")
print(sst.dims)

print("\nShape:")
print(sst.shape)


print("\nDominio real seleccionado:")

print(
    "Longitud:",
    float(sst.longitude.min()),
    "a",
    float(sst.longitude.max())
)

print(
    "Latitud :",
    float(sst.latitude.min()),
    "a",
    float(sst.latitude.max())
)


print("\nRango SST:")

print(
    float(
        sst.min(
            skipna=True
        )
    ),
    "a",
    float(
        sst.max(
            skipna=True
        )
    ),
    "°C"
)


# ============================================================
# 11. CONTROL TEMPORAL
# ============================================================

N_DIAS_ESPERADOS = (
    N_SEMANAS
    *
    DIAS_SEMANA
)

N_DIAS_DISPONIBLES = (
    sst.sizes[
        "lead_time"
    ]
)


print("\n" + "=" * 100)
print("CONTROL TEMPORAL")
print("=" * 100)

print(
    "Días esperados   :",
    N_DIAS_ESPERADOS
)

print(
    "Días disponibles :",
    N_DIAS_DISPONIBLES
)


if (
    N_DIAS_DISPONIBLES
    <
    N_DIAS_ESPERADOS
):

    raise ValueError(
        f"No existen suficientes días. "
        f"Disponibles={N_DIAS_DISPONIBLES}, "
        f"necesarios={N_DIAS_ESPERADOS}"
    )


print(
    "\nOK: existen las 6 semanas completas."
)


# ============================================================
# 12. CALCULAR PROMEDIOS SEMANALES
#
# Paso 1:
#   promedio de 7 días para cada miembro
#
# Paso 2:
#   promedio de los miembros
# ============================================================

sst_semanal = {}

sst_semanal_miembros = {}

fechas_semanas = {}


print("\n" + "=" * 100)
print("PROMEDIOS SEMANALES")
print("=" * 100)


for semana in range(
    1,
    N_SEMANAS + 1
):


    # ========================================================
    # ÍNDICES TEMPORALES
    # ========================================================

    i0 = (
        (semana - 1)
        *
        DIAS_SEMANA
    )

    i1 = (
        i0
        +
        DIAS_SEMANA
    )


    # ========================================================
    # EXTRAER LOS 7 DÍAS
    # ========================================================

    datos_semana = (
        sst.isel(
            lead_time=slice(
                i0,
                i1
            )
        )
    )


    # ========================================================
    # VERIFICAR 7 DÍAS
    # ========================================================

    if (
        datos_semana.sizes[
            "lead_time"
        ]
        != DIAS_SEMANA
    ):

        raise ValueError(
            f"Semana {semana}: "
            f"solo tiene "
            f"{datos_semana.sizes['lead_time']} días."
        )


    # ========================================================
    # PROMEDIO DE LOS 7 DÍAS PARA CADA MIEMBRO
    # ========================================================

    promedio_miembros = (
        datos_semana.mean(
            dim="lead_time",
            skipna=True
        )
    )


    sst_semanal_miembros[
        semana
    ] = promedio_miembros


    # ========================================================
    # PROMEDIO DE LOS 101 MIEMBROS
    # ========================================================

    promedio_ensamble = (
        promedio_miembros.mean(
            dim="ensemble_member",
            skipna=True
        )
    )


    # ========================================================
    # RENOMBRAR LAT/LON
    # ========================================================

    promedio_ensamble = (
        promedio_ensamble.rename(
            {
                "latitude": "lat",
                "longitude": "lon"
            }
        )
    )


    promedio_ensamble = (
        promedio_ensamble.sortby(
            "lat"
        )
    )


    promedio_ensamble = (
        promedio_ensamble.sortby(
            "lon"
        )
    )


    sst_semanal[
        semana
    ] = promedio_ensamble


    # ========================================================
    # FECHAS
    # ========================================================

    if (
        "valid_time"
        in datos_semana.coords
    ):

        fechas = pd.to_datetime(
            datos_semana[
                "valid_time"
            ].values
        )

        fecha_ini = pd.Timestamp(
            fechas[0]
        )

        fecha_fin = pd.Timestamp(
            fechas[-1]
        )


    else:

        lead_ini = pd.Timedelta(
            datos_semana[
                "lead_time"
            ].values[0]
        )

        lead_fin = pd.Timedelta(
            datos_semana[
                "lead_time"
            ].values[-1]
        )

        fecha_ini = (
            INIT
            +
            lead_ini
        )

        fecha_fin = (
            INIT
            +
            lead_fin
        )


    fechas_semanas[
        semana
    ] = (
        fecha_ini,
        fecha_fin
    )


    # ========================================================
    # ESTADÍSTICAS
    # ========================================================

    minimo = float(
        promedio_ensamble.min(
            skipna=True
        )
    )

    maximo = float(
        promedio_ensamble.max(
            skipna=True
        )
    )

    media = float(
        promedio_ensamble.mean(
            skipna=True
        )
    )


    print(
        f"Semana {semana} | "
        f"{fecha_ini:%d-%m-%Y} al "
        f"{fecha_fin:%d-%m-%Y} | "
        f"MIN={minimo:.2f} °C | "
        f"MAX={maximo:.2f} °C | "
        f"MEDIA={media:.2f} °C"
    )


# ============================================================
# 13. PALETA SST
# ============================================================

sst_colors = [

    "#00079F",
    "#0116C8",
    "#0125F0",
    "#013DFF",

    "#015AFE",
    "#0177FE",
    "#2699FE",
    "#58BCFE",

    "#8AE0FE",
    "#A8EAFE",
    "#C3F0FF",
    "#DEF7FF",

    "#FFF6C1",
    "#FFF19D",
    "#FFDE5D",
    "#FFC50E",

    "#FF9501",
    "#FF6101",
    "#F62A00",
    "#E72200",

    "#D71A00",
    "#C81101",
    "#B80901",
    "#A90100",
]


# ============================================================
# 14. ESCALA SST DISCRETA
# 16, 17, 18 ... 32 °C
# ============================================================

SST_MIN = 16

SST_MAX = 32

SST_INTERVALO = 1


SST_LEVELS = np.arange(
    SST_MIN,
    SST_MAX
    +
    SST_INTERVALO,
    SST_INTERVALO
)


# ============================================================
# 15. CREAR PALETA DISCRETA
# ============================================================

sst_cmap_base = (
    mcolors.LinearSegmentedColormap.from_list(
        "sst_base",
        sst_colors,
        N=256
    )
)


sst_cmap = (
    mcolors.ListedColormap(
        sst_cmap_base(
            np.linspace(
                0,
                1,
                len(SST_LEVELS) - 1
            )
        ),
        name="sst_discreta"
    )
)


# ============================================================
# NORMALIZACIÓN
# ============================================================

sst_norm = (
    mcolors.BoundaryNorm(
        boundaries=SST_LEVELS,
        ncolors=sst_cmap.N,
        clip=True
    )
)


# ============================================================
# 16. ISOTERMAS
#
# 26 °C
# 27 °C
# 28.5 °C
# 29 °C
# ============================================================

contour_levels = [

    26.0,

    27.0,

    28.5,

    29.0
]


contour_colors = [

    "limegreen",      # 26 °C

    "purple",         # 27 °C

    "gold",           # 28.5 °C

    "deepskyblue"     # 29 °C
]


GROSOR_ISOTERMAS = 4.0

TAMANO_ETIQUETA_ISOTERMA = 10


# ============================================================
# 17. FUNCIÓN GEOGRAFÍA
# SIN TOPOGRAFÍA
# ============================================================

def agregar_geografia(
    ax
):

    # --------------------------------------------------------
    # TIERRA
    # --------------------------------------------------------

    ax.add_feature(
        cfeature.LAND,
        facecolor="#EEEEEE",
        edgecolor="none",
        zorder=8
    )


    # --------------------------------------------------------
    # COSTA
    # --------------------------------------------------------

    ax.add_feature(
        cfeature.COASTLINE,
        linewidth=0.9,
        edgecolor="black",
        zorder=12
    )


    # --------------------------------------------------------
    # FRONTERAS
    # --------------------------------------------------------

    ax.add_feature(
        cfeature.BORDERS,
        linewidth=0.45,
        edgecolor="black",
        zorder=12
    )


# ============================================================
# 18. FUNCIÓN PARA ISOTERMAS
# ============================================================

def agregar_isotermas(
    ax,
    campo
):

    try:

        cs = ax.contour(

            campo.lon,

            campo.lat,

            campo,

            levels=
            contour_levels,

            colors=
            contour_colors,

            linewidths=
            GROSOR_ISOTERMAS,

            transform=
            ccrs.PlateCarree(),

            zorder=7
        )


        # ====================================================
        # ETIQUETAS
        # ====================================================

        textos = ax.clabel(

            cs,

            fmt={
                26.0: "26°C",
                27.0: "27°C",
                28.5: "28.5°C",
                29.0: "29°C"
            },

            inline=True,

            fontsize=
            TAMANO_ETIQUETA_ISOTERMA
        )


        # ====================================================
        # NEGRITA
        # ====================================================

        for texto in textos:

            texto.set_fontweight(
                "bold"
            )

            texto.set_bbox(
                None
            )


    except Exception as e:

        print(
            "Advertencia al dibujar isotermas:",
            e
        )


# ============================================================
# 19. CONFIGURAR MAPA
# ============================================================

def configurar_mapa(
    ax,
    mostrar_latitud=False
):

    # --------------------------------------------------------
    # DOMINIO
    # --------------------------------------------------------

    ax.set_extent(
        [
            LON_MIN,
            LON_MAX,
            LAT_MIN,
            LAT_MAX
        ],
        crs=ccrs.PlateCarree()
    )


    # --------------------------------------------------------
    # GRILLA
    # --------------------------------------------------------

    gl = ax.gridlines(

        draw_labels=True,

        crs=ccrs.PlateCarree(),

        linestyle="--",

        linewidth=0.30,

        alpha=0.45
    )


    gl.top_labels = False

    gl.right_labels = False


    gl.left_labels = (
        mostrar_latitud
    )


    gl.xlabel_style = {
        "size": 8
    }


    gl.ylabel_style = {
        "size": 8
    }


# ============================================================
# 20. CREAR FIGURA
#
# 1 FILA x 6 SEMANAS
# ============================================================

fig, axs = plt.subplots(

    1,

    6,

    figsize=(
        25,
        6.5
    ),

    subplot_kw={
        "projection":
        ccrs.PlateCarree()
    }
)


# ============================================================
# TÍTULO GENERAL
# ============================================================

fig.suptitle(

    "Pronóstico Semanal de Temperatura Superficial del Mar\n"
    "ENS- 101 Simulaciones",

    fontsize=18,

    fontweight="bold",

    y=0.96
)


im = None


# ============================================================
# 21. GRAFICAR LAS 6 SEMANAS
# ============================================================

for i in range(
    N_SEMANAS
):


    semana = (
        i + 1
    )


    ax = axs[i]


    # ========================================================
    # CONFIGURACIÓN DEL MAPA
    # ========================================================

    configurar_mapa(

        ax,

        mostrar_latitud=(
            i == 0
        )
    )


    campo = (
        sst_semanal[
            semana
        ]
    )


    # ========================================================
    # SST
    # ========================================================

    im = ax.pcolormesh(

        campo.lon,

        campo.lat,

        campo,

        cmap=
        sst_cmap,

        norm=
        sst_norm,

        shading="auto",

        transform=
        ccrs.PlateCarree(),

        zorder=1
    )


    # ========================================================
    # ISOTERMAS
    # ========================================================

    agregar_isotermas(
        ax,
        campo
    )


    # ========================================================
    # GEOGRAFÍA
    # ========================================================

    agregar_geografia(
        ax
    )


    # ========================================================
    # FECHAS
    # ========================================================

    fecha_ini, fecha_fin = (
        fechas_semanas[
            semana
        ]
    )


    # ========================================================
    # TÍTULO DEL PANEL
    # ========================================================

    ax.set_title(

        f"SEMANA {semana}\n"
        f"{fecha_ini:%d-%m-%Y} al "
        f"{fecha_fin:%d-%m-%Y}",

        fontsize=10,

        fontweight="bold",

        pad=7
    )


# ============================================================
# 22. ESPACIADO
# ============================================================

plt.subplots_adjust(

    left=0.055,

    right=0.975,

    bottom=0.25,

    top=0.82,

    wspace=0.10
)


# ============================================================
# 23. COLORBAR SST
# ============================================================

if im is not None:


    cbar_ax = fig.add_axes(
        [
            0.20,
            0.095,
            0.60,
            0.035
        ]
    )


    cbar = fig.colorbar(

        im,

        cax=cbar_ax,

        orientation="horizontal",

        boundaries=SST_LEVELS,

        ticks=SST_LEVELS,

        spacing="uniform",

        drawedges=True
    )


    cbar.set_label(

        "Temperatura Superficial del Mar (°C)",

        fontsize=13,

        fontweight="bold",

        labelpad=8
    )


    cbar.ax.tick_params(

        axis="x",

        labelsize=10,

        length=4,

        width=0.9,

        pad=3
    )


    cbar.ax.set_xticklabels(
        [
            f"{int(v)}"
            for v in SST_LEVELS
        ]
    )


    for tick in (
        cbar.ax.get_xticklabels()
    ):

        tick.set_fontweight(
            "bold"
        )


    cbar.dividers.set_linewidth(
        0.45
    )


# ============================================================
# 24. LEYENDA DE ISOTERMAS
# ============================================================

legend_elements = [


    # --------------------------------------------------------
    # 26 °C
    # --------------------------------------------------------

    Line2D(
        [0],
        [0],

        color="limegreen",

        lw=5,

        label="26°C"
    ),


    # --------------------------------------------------------
    # 27 °C
    # --------------------------------------------------------

    Line2D(
        [0],
        [0],

        color="purple",

        lw=5,

        label="27°C"
    ),


    # --------------------------------------------------------
    # 28.5 °C
    # --------------------------------------------------------

    Line2D(
        [0],
        [0],

        color="gold",

        lw=5,

        label="28.5°C"
    ),


    # --------------------------------------------------------
    # 29 °C
    # --------------------------------------------------------

    Line2D(
        [0],
        [0],

        color="deepskyblue",

        lw=5,

        label="29°C"
    )

]


leyenda = fig.legend(

    handles=
    legend_elements,

    loc=
    "lower center",

    bbox_to_anchor=(
        0.5,
        0.15
    ),

    ncol=4,

    fontsize=10,

    frameon=False,

    title=
    "ISOTERMAS",

    handlelength=3
)


plt.setp(

    leyenda.get_texts(),

    fontweight="bold"
)


leyenda.get_title().set_fontweight(
    "bold"
)


# ============================================================
# 25. INFORMACIÓN INFERIOR
# ============================================================



# ============================================================
# 26. GUARDAR FIGURA
# ============================================================

fecha_init_str = (
    INIT.strftime(
        "%Y%m%d"
    )
)


OUTPUT_FILE = os.path.join(

    OUTPUT_DIR,

    (
        f"ocen_ecmwf_ifs_"
        f"{fecha_init_str}_"
        f"sst_semanal.png"
    )
)


plt.savefig(

    OUTPUT_FILE,

    dpi=300,

    bbox_inches="tight",

    facecolor="white"
)


# ============================================================
# 27. INFORMACIÓN FINAL
# ============================================================

print("\n" + "=" * 100)

print(
    "FIGURA GUARDADA:"
)

print(
    OUTPUT_FILE
)

print("=" * 100)


print("\nSEMANAS GENERADAS:")

for semana in range(
    1,
    N_SEMANAS + 1
):

    fecha_ini, fecha_fin = (
        fechas_semanas[
            semana
        ]
    )

    print(
        f"Semana {semana}: "
        f"{fecha_ini:%d-%m-%Y} "
        f"al "
        f"{fecha_fin:%d-%m-%Y}"
    )


# ============================================================
# 28. MOSTRAR FIGURA
# ============================================================

plt.show()
```

```python

# -*- coding: utf-8 -*-
# ================================================================
# ECMWF IFS ENS | PRONÓSTICO DE PWAT
# HOVMÖLLER COSTERO DEL PERÚ
#
# 42 DÍAS | D+2 A D+43
#
# SALIDAS:
# 1. HOVMÖLLER MAR + SERIE COSTA NORTE + MAPA GIRADO
# 2. HOVMÖLLER TIERRA + SERIE COSTA NORTE + MAPA GIRADO
# 3. CSV DE SERIES DIARIAS
#
# SIN INTERPOLACIÓN NI SUAVIZADO
# CRÉDITOS: SENAMHI/DSCA/SCM
# ================================================================

# ================================================================
# 0. INSTALACIÓN
# ================================================================

%pip install -q dynamical-catalog cartopy shapely pyproj netcdf4

# ================================================================
# 1. LIBRERÍAS
# ================================================================

import os
import numpy as np
import pandas as pd
import xarray as xr

import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib.dates as mdates

from matplotlib.lines import Line2D

import cartopy.crs as ccrs
import cartopy.feature as cfeature
from cartopy.io import shapereader

from shapely.geometry import Point, Polygon, box
from shapely.ops import unary_union, transform
from shapely.prepared import prep

from pyproj import Transformer

import dynamical_catalog

# ================================================================
# 2. CONFIGURACIÓN
# ================================================================

CATALOGO = (
    "ecmwf-ifs-ens-forecast-46-day-daily-1-5-degree"
)

VARIABLES_CANDIDATAS = [
    "total_column_water_vapour",
    "total_column_water_vapor",
    "atmosphere_mass_content_of_water_vapor",
    "precipitable_water",
    "total_column_water_atmosphere"
]

N_DIAS = 42
DIA_INICIAL = 2

DISTANCIA_COSTA_KM = 150

# Dominio de extracción
LON_MIN, LON_MAX = -84, -68
LAT_MIN, LAT_MAX = -21, 1

# Hovmöller
LAT_SUR = -18.3
LAT_NORTE = -3.5
PASO_LAT = 1.5

# Costa norte
LAT_NORTE_MIN = -9.0
LAT_NORTE_MAX = -4.0

# Umbrales
UMBRAL_AMARILLO = 50.0
UMBRAL_ROJO = 55.0

COLOR_AZUL = "#0876B9"
COLOR_AMARILLO = "#F2B600"
COLOR_ROJO = "#DC2626"

# Directorio
OUTPUT_DIR = "/content/HOVMOLLER_COSTERO_ECMWF"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

# ================================================================
# 3. ABRIR CATÁLOGO ECMWF
# ================================================================

print("=" * 80)
print("ECMWF IFS ENS - PRONÓSTICO COSTERO DE PWAT")
print("=" * 80)

ds = dynamical_catalog.open(
    CATALOGO,
    chunks=None
)

print("\nDimensiones:")
print(ds.sizes)

# ================================================================
# 4. IDENTIFICAR VARIABLE
# ================================================================

VAR_PW = next(
    (
        variable
        for variable in VARIABLES_CANDIDATAS
        if variable in ds.data_vars
    ),
    None
)

if VAR_PW is None:
    raise KeyError(
        "No se encontró variable compatible. "
        f"Variables disponibles: {list(ds.data_vars)}"
    )

print("\nVariable seleccionada:", VAR_PW)
print("Atributos:", ds[VAR_PW].attrs)

if VAR_PW == "total_column_water_atmosphere":
    print(
        "\nAVISO TÉCNICO: se utiliza TCW. "
        "TCW no es estrictamente igual a PWAT/TCWV."
    )

# ================================================================
# 5. INICIALIZACIÓN
# ================================================================

INIT_TIME = ds.init_time.values[-1]
INIT = pd.Timestamp(INIT_TIME)

N_MIEMBROS = ds.sizes["ensemble_member"]

print("\nInicialización:", INIT)
print("Miembros:", N_MIEMBROS)

# ================================================================
# 6. SELECCIONAR DOMINIO
# ================================================================

pw = ds[VAR_PW].sel(
    init_time=INIT_TIME
)

if float(pw.longitude.max()) > 180:

    pw = pw.assign_coords(
        longitude=(
            (pw.longitude + 180) % 360
        ) - 180
    )

pw = pw.sortby("longitude")
pw = pw.sortby("latitude")

pw = pw.sel(
    longitude=slice(LON_MIN, LON_MAX),
    latitude=slice(LAT_MIN, LAT_MAX)
)

# ================================================================
# 7. SELECCIONAR 42 DÍAS
# ================================================================

lead_inicio = pd.Timedelta(
    days=DIA_INICIAL
)

lead_fin = pd.Timedelta(
    days=DIA_INICIAL + N_DIAS - 1
)

pw = pw.sel(
    lead_time=slice(
        lead_inicio,
        lead_fin
    )
)

lead = pd.to_timedelta(
    pw.lead_time.values
)

lead_esperado = pd.timedelta_range(
    start=lead_inicio,
    periods=N_DIAS,
    freq="D"
)

if not np.array_equal(
    lead.values.astype("timedelta64[ns]"),
    lead_esperado.values.astype("timedelta64[ns]")
):
    raise ValueError(
        "Los 42 pasos diarios no están completos."
    )

fechas = pd.DatetimeIndex(
    INIT + lead
)

print(
    "\nPeriodo:",
    fechas[0].strftime("%d/%m/%Y"),
    "a",
    fechas[-1].strftime("%d/%m/%Y")
)

# ================================================================
# 8. CONVERSIÓN DE UNIDADES
# ================================================================

unidad = str(
    pw.attrs.get("units", "")
).lower().replace(" ", "")

if unidad in (
    "kgm-2",
    "kg/m2",
    "kgm**-2",
    "kgm^-2",
    "mm"
):
    factor = 1.0

elif unidad in (
    "m",
    "meter",
    "metre"
):
    factor = 1000.0

else:
    raise ValueError(
        f"Unidad no reconocida: {unidad}"
    )

# ================================================================
# 9. PROMEDIO DEL ENSAMBLE
# ================================================================

print("\nCalculando promedio del ensamble...")

pw_media = (
    pw.mean(
        dim="ensemble_member",
        skipna=True
    ) * factor
)

pw_media = pw_media.transpose(
    "lead_time",
    "latitude",
    "longitude"
).load()

print("Promedio calculado correctamente.")

# ================================================================
# 10. GEOMETRÍA DEL PERÚ
# ================================================================

print("\nCargando geometrías...")

ruta_paises = shapereader.natural_earth(
    resolution="10m",
    category="cultural",
    name="admin_0_countries"
)

reader_paises = shapereader.Reader(
    ruta_paises
)

peru = None

for registro in reader_paises.records():

    nombre = (
        registro.attributes.get("ADMIN")
        or registro.attributes.get("NAME")
    )

    if nombre == "Peru":
        peru = registro.geometry
        break

if peru is None:
    raise ValueError(
        "No se encontró Perú."
    )

# ================================================================
# 11. TIERRA REGIONAL
# ================================================================

ruta_tierra = shapereader.natural_earth(
    resolution="10m",
    category="physical",
    name="land"
)

reader_tierra = shapereader.Reader(
    ruta_tierra
)

dominio_regional = box(
    -90, -25, -65, 5
)

partes_tierra = []

for geom in reader_tierra.geometries():

    if geom.intersects(
        dominio_regional
    ):

        parte = geom.intersection(
            dominio_regional
        )

        if not parte.is_empty:
            partes_tierra.append(parte)

tierra_regional = unary_union(
    partes_tierra
)

# ================================================================
# 12. PROYECCIÓN MÉTRICA
# ================================================================

CRS_METRICO = (
    "+proj=aeqd "
    "+lat_0=-11 "
    "+lon_0=-77 "
    "+datum=WGS84 "
    "+units=m "
    "+no_defs"
)

transformador = Transformer.from_crs(
    "EPSG:4326",
    CRS_METRICO,
    always_xy=True
)

a_metros = transformador.transform

peru_m = transform(
    a_metros,
    peru
)

# ================================================================
# 13. COSTA PERUANA
# ================================================================

ruta_costas = shapereader.natural_earth(
    resolution="10m",
    category="physical",
    name="coastline"
)

reader_costas = shapereader.Reader(
    ruta_costas
)

costas_candidatas = []

for geom in reader_costas.geometries():

    if geom.intersects(
        peru.buffer(0.2)
    ):
        costas_candidatas.append(geom)

if not costas_candidatas:
    raise ValueError(
        "No se encontraron líneas de costa."
    )

costas_union = unary_union(
    costas_candidatas
)

costas_m = transform(
    a_metros,
    costas_union
)

costa_peru_m = costas_m.intersection(
    peru_m.buffer(15000)
)

if costa_peru_m.is_empty:
    raise ValueError(
        "No se pudo determinar la costa peruana."
    )

print("Costa peruana obtenida.")

# ================================================================
# 14. CLASIFICACIÓN DE CELDAS
# ================================================================

lats = pw_media.latitude.values
lons = pw_media.longitude.values

LON2D, LAT2D = np.meshgrid(
    lons,
    lats
)

puntos_lon = LON2D.ravel()
puntos_lat = LAT2D.ravel()

n_puntos = len(puntos_lon)

peru_preparado = prep(peru)
tierra_preparada = prep(tierra_regional)

distancias_km = np.full(
    n_puntos,
    np.nan
)

es_tierra_peru = np.zeros(
    n_puntos,
    dtype=bool
)

es_oceano = np.zeros(
    n_puntos,
    dtype=bool
)

print("\nClasificando celdas ECMWF...")

for i, (lon, lat) in enumerate(
    zip(puntos_lon, puntos_lat)
):

    punto = Point(
        float(lon),
        float(lat)
    )

    punto_m = transform(
        a_metros,
        punto
    )

    distancias_km[i] = (
        punto_m.distance(
            costa_peru_m
        ) / 1000.0
    )

    es_tierra_peru[i] = (
        peru_preparado.covers(punto)
    )

    es_oceano[i] = (
        not tierra_preparada.covers(punto)
    )

# ================================================================
# 15. MÁSCARAS MAR Y TIERRA
# ================================================================

en_latitud = (
    (puntos_lat >= LAT_SUR)
    & (puntos_lat <= LAT_NORTE)
)

mascara_mar = (
    es_oceano
    & en_latitud
    & (
        distancias_km <=
        DISTANCIA_COSTA_KM
    )
)

mascara_tierra = (
    es_tierra_peru
    & en_latitud
    & (
        distancias_km <=
        DISTANCIA_COSTA_KM
    )
)

print("\nCeldas mar:", int(mascara_mar.sum()))
print("Celdas tierra:", int(mascara_tierra.sum()))

if not mascara_mar.any():
    raise ValueError(
        "No existen celdas marinas seleccionadas."
    )

if not mascara_tierra.any():
    raise ValueError(
        "No existen celdas terrestres seleccionadas."
    )

# ================================================================
# 16. BANDAS LATITUDINALES
# ================================================================

bordes_lat = np.arange(
    LAT_SUR,
    LAT_NORTE,
    PASO_LAT
)

if bordes_lat[-1] < LAT_NORTE:

    bordes_lat = np.append(
        bordes_lat,
        LAT_NORTE
    )

n_bandas = len(
    bordes_lat
) - 1

# ================================================================
# 17. MATRIZ DIARIA
# ================================================================

valores = np.asarray(
    pw_media.values,
    dtype=float
)

nt = valores.shape[0]

valores_2d = valores.reshape(
    nt,
    -1
)

# ================================================================
# 18. HOVMÖLLER
# ================================================================

def calcular_hovmoller(
    mascara_region
):

    hov = np.full(
        (n_bandas, nt),
        np.nan
    )

    conteo = np.zeros(
        n_bandas,
        dtype=int
    )

    for i in range(n_bandas):

        lat0 = bordes_lat[i]
        lat1 = bordes_lat[i + 1]

        if i == n_bandas - 1:

            en_banda = (
                (puntos_lat >= lat0)
                & (puntos_lat <= lat1)
            )

        else:

            en_banda = (
                (puntos_lat >= lat0)
                & (puntos_lat < lat1)
            )

        seleccion = (
            mascara_region
            & en_banda
        )

        conteo[i] = int(
            seleccion.sum()
        )

        if conteo[i] > 0:

            hov[i, :] = np.nanmean(
                valores_2d[:, seleccion],
                axis=1
            )

    return hov, conteo


HOV_MAR, CONTEO_MAR = calcular_hovmoller(
    mascara_mar
)

HOV_TIERRA, CONTEO_TIERRA = calcular_hovmoller(
    mascara_tierra
)

# ================================================================
# 19. SERIE COSTA NORTE
# ================================================================

def calcular_serie_norte(
    mascara_region
):

    seleccion = (
        mascara_region
        & (puntos_lat >= LAT_NORTE_MIN)
        & (puntos_lat <= LAT_NORTE_MAX)
    )

    n_celdas = int(
        seleccion.sum()
    )

    if n_celdas == 0:

        return (
            np.full(nt, np.nan),
            seleccion,
            0
        )

    serie = np.nanmean(
        valores_2d[:, seleccion],
        axis=1
    )

    return (
        serie,
        seleccion,
        n_celdas
    )


SERIE_MAR, SEL_NORTE_MAR, N_MAR = (
    calcular_serie_norte(
        mascara_mar
    )
)

SERIE_TIERRA, SEL_NORTE_TIERRA, N_TIERRA = (
    calcular_serie_norte(
        mascara_tierra
    )
)

print("\nCosta norte:")
print("Celdas mar:", N_MAR)
print("Celdas tierra:", N_TIERRA)

# ================================================================
# 20. PALETA PWAT
# ================================================================

PW_LEVELS = [
    0, 6, 12, 18, 26,
    38, 50, 55, 60, 70
]

PW_COLOR_POINTS = [
    (0.0,  "#70421D"),
    (3.0,  "#98602F"),
    (6.0,  "#0010A8"),
    (9.0,  "#0010FF"),
    (12.0, "#5553A8"),
    (15.0, "#9995EE"),
    (18.0, "#007E93"),
    (22.0, "#00B6CB"),
    (26.0, "#00E8E8"),
    (30.0, "#008E00"),
    (34.0, "#00CC00"),
    (38.0, "#00FF00"),
    (43.0, "#A8A800"),
    (47.0, "#FFFF00"),
    (50.0, "#E00000"),
    (53.0, "#FF0000"),
    (55.0, "#D000D0"),
    (58.0, "#FF00FF"),
    (60.0, "#888888"),
    (65.0, "#AFAFAF"),
    (70.0, "#D0D0D0")
]

PW_CMAP = mcolors.LinearSegmentedColormap.from_list(
    "PWAT_SENAMHI",
    [
        (v / 70.0, c)
        for v, c in PW_COLOR_POINTS
    ],
    N=1024
)

PW_CMAP.set_bad("#FFFFFF")
PW_CMAP.set_over("#D0D0D0")

PW_NORM = mcolors.Normalize(
    vmin=0,
    vmax=70
)

# ================================================================
# 21. BORDES TEMPORALES
# ================================================================

fechas_num = mdates.date2num(
    fechas.to_pydatetime()
)

bordes_tiempo = np.concatenate([
    [fechas_num[0] - 0.5],
    (
        fechas_num[:-1]
        + fechas_num[1:]
    ) / 2.0,
    [fechas_num[-1] + 0.5]
])

# ================================================================
# 22. SERIE CON COLORES SEGÚN UMBRAL
# ================================================================

def dibujar_serie_umbrales(
    ax,
    fechas,
    serie
):

    x = mdates.date2num(
        pd.DatetimeIndex(
            fechas
        ).to_pydatetime()
    )

    y = np.asarray(
        serie,
        dtype=float
    )

    for i in range(len(y) - 1):

        x0, x1 = x[i], x[i + 1]
        y0, y1 = y[i], y[i + 1]

        if not (
            np.isfinite(y0)
            and np.isfinite(y1)
        ):
            continue

        fracciones = [0.0, 1.0]

        if y1 != y0:

            for umbral in (
                UMBRAL_AMARILLO,
                UMBRAL_ROJO
            ):

                f = (
                    (umbral - y0)
                    / (y1 - y0)
                )

                if 0 < f < 1:
                    fracciones.append(f)

        fracciones = sorted(
            set(fracciones)
        )

        for j in range(
            len(fracciones) - 1
        ):

            f0 = fracciones[j]
            f1 = fracciones[j + 1]

            xa = x0 + f0 * (x1 - x0)
            xb = x0 + f1 * (x1 - x0)

            ya = y0 + f0 * (y1 - y0)
            yb = y0 + f1 * (y1 - y0)

            valor_medio = (
                ya + yb
            ) / 2.0

            if valor_medio < 50:
                color = COLOR_AZUL

            elif valor_medio < 55:
                color = COLOR_AMARILLO

            else:
                color = COLOR_ROJO

            ax.plot(
                [xa, xb],
                [ya, yb],
                color=color,
                linewidth=2.7,
                solid_capstyle="round",
                zorder=4
            )

    # Marcadores diarios
    for fecha, valor in zip(
        fechas,
        y
    ):

        if not np.isfinite(valor):
            continue

        if valor < 50:
            color = COLOR_AZUL

        elif valor < 55:
            color = COLOR_AMARILLO

        else:
            color = COLOR_ROJO

        ax.scatter(
            fecha,
            valor,
            color=color,
            edgecolor="white",
            linewidth=0.5,
            s=24,
            zorder=6
        )

# ================================================================
# 23. MAPA REFERENCIAL GIRADO
# ================================================================

def dibujar_mapa_referencial(
    ax,
    region
):

    # Enfoque exclusivo en la costa norte
    ax.set_extent(
        [-83.5, -76.0, -12.0, -2.5],
        crs=ccrs.PlateCarree()
    )

    ax.add_feature(
        cfeature.OCEAN,
        facecolor="#EAF4FA",
        zorder=0
    )

    ax.add_feature(
        cfeature.LAND,
        facecolor="#F0EFEA",
        zorder=0
    )

    ax.coastlines(
        resolution="10m",
        linewidth=1.0,
        color="#333333",
        zorder=6
    )

    ax.add_feature(
        cfeature.BORDERS,
        linewidth=0.5,
        edgecolor="#777777",
        zorder=6
    )

    # ------------------------------------------------------------
    # CAJA INCLINADA REFERENCIAL
    # ------------------------------------------------------------

    caja_costera = Polygon([
        (-83.5, -4.5),
        (-81.4, -3.8),
        (-78.3, -9.0),
        (-80.5, -9.8)
    ])

    ax.add_geometries(
        [caja_costera],
        crs=ccrs.PlateCarree(),
        facecolor="#FF0000",
        edgecolor="#D32F2F",
        linewidth=1.2,
        alpha=0.45,
        zorder=4
    )

    ax.set_title(
        (
            "Área de cálculo de PWAT\n"
            "Costa norte (4°S–9°S)"
        ),
        fontsize=10.5,
        fontweight="bold",
        pad=9
    )

    gl = ax.gridlines(
        draw_labels=True,
        linewidth=0.35,
        color="gray",
        alpha=0.30,
        linestyle="--"
    )

    gl.top_labels = False
    gl.right_labels = False

    gl.xlabel_style = {"size": 8}
    gl.ylabel_style = {"size": 8}

# ================================================================
# 24. FUNCIÓN PARA GENERAR FIGURA
# ================================================================

def generar_figura(
    HOV,
    SERIE,
    REGION,
    ARCHIVO
):

    fig = plt.figure(
        figsize=(17, 8.5),
        facecolor="white"
    )

    gs = fig.add_gridspec(
        nrows=2,
        ncols=2,
        width_ratios=[4.0, 1.7],
        height_ratios=[3.0, 1.05],
        left=0.07,
        right=0.97,
        top=0.79,
        bottom=0.14,
        wspace=0.12,
        hspace=0.18
    )

    ax_hov = fig.add_subplot(
        gs[0, 0]
    )

    ax_serie = fig.add_subplot(
        gs[1, 0],
        sharex=ax_hov
    )

    # ------------------------------------------------------------
    # PROYECCIÓN GIRADA DEL MAPA
    # ------------------------------------------------------------

    proyeccion_mapa = ccrs.TransverseMercator(
        central_longitude=-79,
        central_latitude=-7
    )

    ax_mapa = fig.add_subplot(
        gs[:, 1],
        projection=proyeccion_mapa
    )

    # ------------------------------------------------------------
    # HOVMÖLLER
    # ------------------------------------------------------------

    im = ax_hov.pcolormesh(
        bordes_tiempo,
        bordes_lat,
        HOV,
        cmap=PW_CMAP,
        norm=PW_NORM,
        shading="flat",
        rasterized=True
    )

    ax_hov.set_ylim(
        LAT_SUR,
        LAT_NORTE
    )

    ax_hov.set_yticks(
        [-18, -16, -14, -12, -10, -8, -6, -4]
    )

    ax_hov.set_yticklabels(
        [
            "18°S", "16°S", "14°S", "12°S",
            "10°S", "8°S", "6°S", "4°S"
        ]
    )

    ax_hov.set_ylabel(
        "Latitud costera",
        fontsize=11,
        fontweight="bold"
    )

    ax_hov.axhline(
        -9,
        color="white",
        linestyle="--",
        linewidth=1.4
    )

    ax_hov.tick_params(
        labelbottom=False,
        labelsize=9
    )

    # ------------------------------------------------------------
    # SERIE COSTA NORTE
    # ------------------------------------------------------------

    dibujar_serie_umbrales(
        ax_serie,
        fechas,
        SERIE
    )

    ax_serie.axhline(
        50,
        color=COLOR_AMARILLO,
        linestyle="--",
        linewidth=1.3
    )

    ax_serie.axhline(
        55,
        color=COLOR_ROJO,
        linestyle="--",
        linewidth=1.3
    )

    ax_serie.set_ylim(
        40,
        60
    )

    ax_serie.set_yticks(
        [40, 45, 50, 55, 60]
    )

    ax_serie.set_ylabel(
        "PWAT (mm)",
        fontsize=10,
        fontweight="bold"
    )

    ax_serie.set_title(
        "COSTA NORTE (4°S–9°S)",
        loc="left",
        fontsize=10.5,
        fontweight="bold",
        pad=5
    )

    ax_serie.set_xlabel(
        "Pronóstico diario",
        fontsize=10,
        fontweight="bold"
    )

    ax_serie.xaxis.set_major_locator(
        mdates.DayLocator(interval=4)
    )

    ax_serie.xaxis.set_major_formatter(
        mdates.DateFormatter("%d/%m")
    )

    ax_serie.tick_params(
        axis="x",
        labelrotation=45,
        labelsize=8
    )

    ax_serie.grid(
        linestyle="--",
        linewidth=0.45,
        alpha=0.25
    )

    leyenda = [
        Line2D(
            [0], [0],
            color=COLOR_AZUL,
            linewidth=2.5,
            label="<50 mm"
        ),
        Line2D(
            [0], [0],
            color=COLOR_AMARILLO,
            linewidth=2.5,
            label="50–<55 mm"
        ),
        Line2D(
            [0], [0],
            color=COLOR_ROJO,
            linewidth=2.5,
            label="≥55 mm"
        )
    ]

    ax_serie.legend(
        handles=leyenda,
        loc="upper right",
        ncol=3,
        fontsize=8,
        frameon=False
    )

    # ------------------------------------------------------------
    # MAPA GIRADO
    # ------------------------------------------------------------

    dibujar_mapa_referencial(
        ax_mapa,
        REGION
    )

    # ------------------------------------------------------------
    # TÍTULO GENERAL
    # ------------------------------------------------------------

    fecha_inicio = fechas[0].strftime(
        "%d/%m/%Y"
    )

    fecha_fin = fechas[-1].strftime(
        "%d/%m/%Y"
    )

    fig.suptitle(
        (
            "Pronóstico de PWAT | "
            f"{fecha_inicio} – {fecha_fin}"
        ),
        fontsize=19,
        fontweight="bold",
        y=0.89
    )

    # ------------------------------------------------------------
    # BARRA DE COLORES SOBRE HOVMÖLLER
    # ------------------------------------------------------------

    posicion = ax_hov.get_position()

    ancho_barra = (
        posicion.width * 0.53
    )

    x_barra = (
        posicion.x0
        + (
            posicion.width
            - ancho_barra
        ) / 2
    )

    y_barra = (
        posicion.y1 + 0.028
    )

    cax = fig.add_axes([
        x_barra,
        y_barra,
        ancho_barra,
        0.013
    ])

    cbar = fig.colorbar(
        im,
        cax=cax,
        orientation="horizontal",
        ticks=PW_LEVELS,
        extend="max"
    )

    cbar.ax.tick_params(
        labelsize=7,
        length=2,
        pad=2
    )

    # ------------------------------------------------------------
    # CRÉDITOS
    # ------------------------------------------------------------

 #   fig.text(
 #      0.5,
  #      0.045,
  #      (
  #          "ECMWF IFS ENS 1.5°"
   #         " | Créditos: SENAMHI/DSCA/SCM"
   #    ),
   #     ha="center",
   #    fontsize=8.5,
   #    fontweight="bold"
   #)

    # ------------------------------------------------------------
    # GUARDAR PNG
    # ------------------------------------------------------------

    ruta = os.path.join(
        OUTPUT_DIR,
        ARCHIVO
    )

    fig.savefig(
        ruta,
        dpi=300,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.show()
    plt.close(fig)

    print("\nPNG GUARDADO:", ruta)

    return ruta

# ================================================================
# 25. GENERAR PNG MAR Y TIERRA
# ================================================================

FECHA_STR = INIT.strftime(
    "%Y%m%d"
)

PNG_MAR = generar_figura(
    HOV=HOV_MAR,
    SERIE=SERIE_MAR,
    REGION="MAR",
    ARCHIVO=(
        f"ECMWF_PWAT_HOV_MAR_NORTE_{FECHA_STR}.png"
    )
)

PNG_TIERRA = generar_figura(
    HOV=HOV_TIERRA,
    SERIE=SERIE_TIERRA,
    REGION="TIERRA",
    ARCHIVO=(
        f"ECMWF_PWAT_HOV_TIERRA_NORTE_{FECHA_STR}.png"
    )
)

# ================================================================
# 26. CSV DE SERIES DIARIAS
# ================================================================

tabla = pd.DataFrame({
    "fecha": fechas,
    "mar_mm": SERIE_MAR,
    "tierra_mm": SERIE_TIERRA
})

for region, columna in [
    ("mar", "mar_mm"),
    ("tierra", "tierra_mm")
]:

    tabla[f"{region}_umbral"] = np.select(
        [
            tabla[columna] >= UMBRAL_ROJO,
            tabla[columna] >= UMBRAL_AMARILLO
        ],
        [
            "ROJO",
            "AMARILLO"
        ],
        default="INFERIOR_50"
    )

ARCHIVO_CSV = os.path.join(
    OUTPUT_DIR,
    f"ECMWF_PWAT_COSTA_NORTE_{FECHA_STR}.csv"
)

tabla.to_csv(
    ARCHIVO_CSV,
    index=False,
    encoding="utf-8-sig"
)

print("\nCSV guardado:", ARCHIVO_CSV)

# ================================================================
# 27. REPORTE DE UMBRALES
# ================================================================

print("\n" + "=" * 80)
print("REPORTE DE UMBRALES")
print("=" * 80)

for nombre, serie in [
    ("OCÉANO PACÍFICO", SERIE_MAR),
    ("TERRITORIO CONTINENTAL", SERIE_TIERRA)
]:

    print("\nREGIÓN:", nombre)

    for umbral in (
        UMBRAL_AMARILLO,
        UMBRAL_ROJO
    ):

        indices = np.flatnonzero(
            np.isfinite(serie)
            & (serie >= umbral)
        )

        print(
            f"Días >= {umbral:.0f} mm:",
            len(indices)
        )

        for j in indices:

            print(
                f"  {fechas[j]:%d/%m/%Y}: "
                f"{serie[j]:.1f} mm"
            )

# ================================================================
# 28. FINAL
# ================================================================

print("\n" + "=" * 80)
print("PROCESAMIENTO COMPLETADO")
print("=" * 80)

print("Variable ECMWF:", VAR_PW)
print("Miembros:", N_MIEMBROS)
print("Días:", N_DIAS)

print("\nCeldas costa norte:")
print("Mar:", N_MAR)
print("Tierra:", N_TIERRA)

print("\nArchivos:")
print(PNG_MAR)
print(PNG_TIERRA)
print(ARCHIVO_CSV)

print("\nSin interpolación ni suavizado.")
```
